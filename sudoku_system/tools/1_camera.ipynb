每次运行会覆盖 `results/manual/` 里的旧文件。运行前先关闭其他用到硬件的 Notebook 内核。

```python
# 让本 Notebook 找到上一级的 sudoku 包
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
from sudoku import config, pipeline

out = config.RESULTS / 'manual'
out.mkdir(parents=True, exist_ok=True)

photo = pipeline.capture(out, show=True)       # 加载 camera.bit，拍一张
cells = pipeline.split_cells(photo, out)       # 找棋盘、切 81 格
print('已切出', len(cells), '格，保存在', out / 'cells')
```

## 检查切出来的 81 格

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(9, 9, figsize=(7, 7))
for ax, cell in zip(axes.ravel(), cells):
    ax.imshow(cell, cmap='gray', vmin=0, vmax=255)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
