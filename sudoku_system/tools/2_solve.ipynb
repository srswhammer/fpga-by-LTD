# 第 2 步：硬件 CNN 识别 + 求解（加载 cnn_solver.bit）

默认读取第 1 步保存的 `results/manual/cells/`。没接摄像头时，可以改用自带的样例
`tests/data/sample_cells/`（见下面代码里的注释）。

结果保存到 `results/manual/lcd_result.json`，第 3 步从这里读。

```python
# 让本 Notebook 找到上一级的 sudoku 包
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
from sudoku import config, grid, lcd, pipeline

out = config.RESULTS / 'manual'
cells_dir = out / 'cells'
# cells_dir = config.ROOT / 'tests' / 'data' / 'sample_cells'   # 没拍照时用样例

cells = grid.load_cells(cells_dir)
result = pipeline.solve(cells)                 # 预处理 → 加载 cnn_solver.bit → 识别 + 求解
```

## 校验并保存给 LCD 的数据

答案不合法（无解、原题被改动、行列宫有重复）会在这里报错，不会保存。
注意：答案合法不代表识别全对，请对照照片核对原题。

```python
payload = lcd.validate_result(result['status'], result['puzzle'], result['answer'], result['given'])
out.mkdir(parents=True, exist_ok=True)
lcd.save_result(out, result['puzzle'], result['answer'], payload, result['steps'], source=cells_dir)
print('已保存：', out / 'lcd_result.json')
```
