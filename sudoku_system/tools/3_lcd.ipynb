# 第 3 步：LCD 显示（加载 lcd.bit）

读取第 2 步保存的 `results/manual/lcd_result.json`，加载 LCD bit 后经电脑桥发送，等待 ACK。

需要电脑上已运行 `pc/lcd_bridge.py`，USB 接 PL UART，网线连好。

```python
# 让本 Notebook 找到上一级的 sudoku 包
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
from sudoku import config, lcd

payload = lcd.load_payload(config.RESULTS / 'manual' / 'lcd_result.json')
lcd.check_bridge()                  # 电脑桥没开会在这里报错
reply = lcd.show(payload)           # 加载 lcd.bit → 发送 → 等 ACK
print('LCD 已确认接收：', reply)
```
